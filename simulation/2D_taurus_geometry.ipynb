```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt
import mc
import seaborn as sns
```

```python
N = 1500
irrational_p = 1/np.sqrt(2)
rational_p = 0.5

iteration_count = 50000
number_runs = 1
```

Condition for different behaviors between rational and irrational probabilities only happens when we obtain a correct solution for $m_1$ and $m_2$ with the argmin function below

```python
mc.find_opt_int(N, rational_p)
```

Simulate $number\_runs$ walks on the taurus, each with $iteration\_count$ steps and with the chosen transition probabilities. Here, the mean of all these runs is computed, as in $ \newline
pass\_count(x,y) = \frac{1}{number\_runs}\sum\limits_{k=1}^{number\_runs}pass\_count_k(x,y)
\newline$

```python
rational_run = mc.compute_path(N, rational_p, number_runs, iteration_count).mean(axis=0)
irrational_run = mc.compute_path(N, irrational_p, number_runs, iteration_count).mean(axis=0)
```

```python
x_index = np.repeat(np.arange(N), N)
y_index = [i for i in range(N)]*N
```

```python
df_rational = pd.DataFrame({"x": x_index, "y": y_index, "count": rational_run})
df_irrational = pd.DataFrame({"x": x_index, "y": y_index, "count": irrational_run})
```

```python
df_rational.to_csv(f"rational_{N}_{number_runs}.csv")
df_irrational.to_csv(f"irrational_{N}_{number_runs}.csv")
```

```python
plt.scatter(x=df_rational["x"], y=df_rational["y"], s=df_rational["count"])
plt.title(f"Random walk with p={rational_p} on {N}x{N} taurus")
```

```python
plt.scatter(x=df_irrational["x"], y=df_irrational["y"], s=df_irrational["count"])
plt.title(f"Random walk with p=1/sqrt(2) on {N}x{N} taurus")
```
